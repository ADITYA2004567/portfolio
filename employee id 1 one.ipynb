```python

def extract_skills(job_description):
    f_skills = []
    job_description = job_description.upper()
    for skill in tech_skills:
        if skill.upper() in job_description:
            f_skills.append(skill)
    return f_skills
```

```python
job_description = """We are looking for a Software Engineer with experience in Python, JavaScript, and SQL.
Familiarity with AWS and Docker is a plus"""
tech_skills = ["pyhton", "java", "reactjs", "sql", "nosql","AWS"]
```

```python
extract_skills(job_description)
```

```python
def valid_email(email):
    if "." in email and "@" in email[-1]:
        return True
    return False

def valid_phnumber(phnumber):
    if len(phnumber) == 10:
        return True
    return False

```

```python
def qualifies_for_job(candidate_skills, job_requirements):
    for skill in job_requirements:
        if skill  not in candidate_skills:
            return False
        return True
```

```python
candidate = ["python", "sql", "machine learning"]
job = ["python"]
print(qualifies_for_job(candidate, job))
```

```python
from datetime import datetime

def calculate_stage(time):
    hiring_stages = [
        'Application Submitted',
        'Application Reviewed',
        'Screening Call Scheduled',
        'Screening Call Completed',
        'Interview Scheduled',
        'Interview Completed',
        'Offer Extended',
        'Offer Accepted',
        'Onboarding Started'
    ]
    durations = {}
    x = list(time.keys())
    for i in range(len(x) - 1):
        s = x[i]
        next_stage = x[i + 1]
        durations[s] = {
            'stage': hiring_stages[i],
            'duration': time[next_stage] - time[s],
            'timestamp': datetime.now().strftime('%Y-%m-%d %H:%M:%S')
        }
    # Calculate total time spent across all stages
    total_time = sum(time[next_stage] - time[s] for s, next_stage in zip(x, x[1:]))
    durations['total_time_spent'] = total_time
    return durations
for stage, info in durations.items():
        if stage != 'total_time_spent':
            print(f"Stage: {info['stage']}, Duration: {info['duration']} seconds, Timestamp: {info['timestamp']}")
print(f"Total Time Spent: {total_time} seconds")
```

```python

def filter_candidates(candidates):
    valid = [c for c in candidates if all(c.get(f) for f in ["name", "email", "phone", "resume"])]
    invalid = [c for c in candidates if c not in valid]
    return valid, invalid
```

```python
candidates = [
    {"name": "adi", "email": "jadi@ee.com", "phone": "9876567887", "resume": "resume.pdf"},
    {"name": "Appu", "email": "aappu@yy.com", "phone": "123", "resume": "resume.pdf"},
    {"name": "sam", "email": "sammm@", "phone": "12345", "resume": ""},
]

valid, invalid = filter_candidates(candidates)
```

```python
print("Valid:", valid)
print("Invalid:", invalid)
```

```python
def assign(candidates, slots, random):
    if len(candidates) > len(slots):
        return "Not enough slots for all candidates."
    random.shuffle(slots)  
    return {candidates[i]: slots[i] for i in range(len(candidates))}
```

```python

candidates = ["Aditya", "don"]
slots = ["10 AM", "12 pM"]
```

```python
assignments = assign(candidates, slots)
print(assignments)
```

```python
def detect_duplicates(emails):
    seen = set()
    duplicates = []
    for email in emails:
        if email in seen:
            duplicates.append(email)
        else:
            seen.add(email)
    
```

```python

emails = [
    "john.doe@gmail.com",
    "johndoe@gmail.com",
    "adi.doe+1@gmail.com",
    "joh23@gmail.com"
]

```

```python
print(detect_duplicates(emails))
```

```python
def sort_candidates(candidates):
    return sorted(candidates, key=lambda x: (-x['experience'], x['expected_salary']))
```

```python
candidates = [
    { "experience": 5, "expected_salary": 70000},
    { "experience": 7, "expected_salary": 90000},
    { "experience": 5, "expected_salary": 65000},
    { "experience": 7, "expected_salary": 85000}
]
```

```python
sorted_candidates = sort_candidates(candidates)
for candidate in sorted_candidates:
    print(candidate)
```
